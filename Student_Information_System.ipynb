```python

import csv
# File name for student information
FILE_NAME = "students.csv"
```

```python

def load_students_from_file():
    """Load student information from the CSV file."""
    students = []
    try:
        with open(FILE_NAME, "r", newline="") as file:
            reader = csv.DictReader(file)
            for row in reader:
                students.append(row)
        print("Student data loaded successfully.")
    except FileNotFoundError:
        print("No student data file found.")
    return students
```

```python

def save_students_to_file(students):
    """Save student information to the CSV file."""
    fields = ["name", "age", "grade1", "grade2",
              "grade3", "average", "category"]

    with open(FILE_NAME, "w", newline="") as file:
        writer = csv.DictWriter(file, fieldnames=fields)
        writer.writeheader()

        for student in students:
            writer.writerow(student)

    print("Student data saved successfully.")
```

```python

def register_student(students):
    """Register a new student and calculate the average grade."""

    name = input("Enter student name: ").strip()

    if name == "":
        print("Student name cannot be empty.")
        return

    try:
        age = int(input("Enter student age: "))

        if age <= 0:
            print("Age must be greater than zero.")
            return

        grade1 = float(input("Enter grade for subject 1: "))
        grade2 = float(input("Enter grade for subject 2: "))
        grade3 = float(input("Enter grade for subject 3: "))

    except ValueError:
        print("Invalid input. Please enter numbers.")
        return

    if not all(0 <= grade <= 100 for grade in [grade1, grade2, grade3]):
        print("Grades must be between 0 and 100.")
        return

    average = (grade1 + grade2 + grade3) / 3

    if average >= 90:
        category = "Excellent"
    elif average >= 70:
        category = "Good"
    else:
        category = "Needs Improvement"

    student = {
        "name": name,
        "age": age,
        "grade1": grade1,
        "grade2": grade2,
        "grade3": grade3,
        "average": round(average, 2),
        "category": category
    }

    students.append(student)

    print(f"{name} has been registered with an average grade of {average:.2f} - {category}")
```

```python

def view_students(students):
    """Display information about all students."""

    if len(students) == 0:
        print("No students found.")
        return

    for student in students:
        print("Name:", student["name"])
        print("Age:", student["age"])
        print("Grade 1:", student["grade1"])
        print("Grade 2:", student["grade2"])
        print("Grade 3:", student["grade3"])
        print("Average Grade:", student["average"])
        print("Category:", student["category"])
        print("--------------------")
```

```python

def search_student(students):
    """Search for a student by name."""

    name = input("Enter student name to search: ").strip()

    found = False

    for student in students:
        if student["name"].lower() == name.lower():
            print("Student found!")
            print("Name:", student["name"])
            print("Age:", student["age"])
            print("Grade 1:", student["grade1"])
            print("Grade 2:", student["grade2"])
            print("Grade 3:", student["grade3"])
            print("Average Grade:", student["average"])
            print("Category:", student["category"])

            found = True
            break

    if found == False:
        print("Student not found.")
```

```python
def update_student(students):
    """Update student information."""

    name = input("Enter student name to update: ").strip()

    for student in students:
        if student["name"].lower() == name.lower():
            print("Student found!")

            new_name = input("Enter new name: ").strip()

            try:
                new_age = int(input("Enter new age: "))
                grade1 = float(input("Enter new grade 1: "))
                grade2 = float(input("Enter new grade 2: "))
                grade3 = float(input("Enter new grade 3: "))
            except ValueError:
                print("Please enter valid numbers.")
                return

            if new_age <= 0:
                print("Age must be greater than 0.")
                return

            if not all(0 <= grade <= 100 for grade in [grade1, grade2, grade3]):
                print("Grades must be between 0 and 100.")
                return

            average = (grade1 + grade2 + grade3) / 3

            if average >= 90:
                category = "Excellent"
            elif average >= 70:
                category = "Good"
            else:
                category = "Needs Improvement"

            student["name"] = new_name
            student["age"] = new_age
            student["grade1"] = grade1
            student["grade2"] = grade2
            student["grade3"] = grade3
            student["average"] = round(average, 2)
            student["category"] = category

            print("Student information updated successfully.")
            return

    print("Student not found.")
```

```python

def delete_student(students):
    """Delete a student by name."""

    name = input("Enter student name to delete: ").strip()

    for student in students:
        if student["name"].lower() == name.lower():
            students.remove(student)
            print("Student deleted successfully.")
            return

    print("Student not found.")
```

```python

def display_menu():
    """Display the main menu options."""

    print("\nStudent Information System")
    print("1. Register a new student")
    print("2. View all students")
    print("3. Search for a student")
    print("4. Update student")
    print("5. delet student")
    print("6. Save and Exit")
    print("7. Exit without Saving")
```

```python

def main():
    """Run the Student Information System."""

    students = load_students_from_file()

    while True:
        display_menu()

        choice = input("Select an option (1-7): ").strip()

        if choice == "1":
            register_student(students)

        elif choice == "2":
            view_students(students)

        elif choice == "3":
            search_student(students)

        elif choice == "4":
            update_student(students)

        
        elif choice == "5":
            delete_student(students)

        elif choice == "6":
            save_students_to_file(students)
            print("Exiting program. Data saved.")
            break

        elif choice == "7":
            print("Exiting without saving.")
            break

        else:
            print("Invalid option. Please choose 1-7.")
```

```python
main()
```
